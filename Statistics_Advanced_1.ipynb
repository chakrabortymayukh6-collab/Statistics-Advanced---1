```python
#Question 1: What is a random variable in probability theory?
```

- A random variable is a variable that assigns a numerical value to each possible outcome of a random experiment

```python
#Question 2: What are the types of random variables?
```

- 1. Discrete Random Variable: Takes countable values, such as 0, 1, 2, 3, etc.
Example: The number of heads obtained when tossing a coin three times.
2. Continuous Random Variable: Takes any value within a given range.
Example: The height, weight, or temperature of a person or place

```python
#Question 3: Explain the difference between discrete and continuous distributions.
```

- discrete and continuous distributions.
1. Discrete Distribution: Represents a random variable that takes countable, distinct values. It uses a probability mass function (PMF).

Example: The number obtained when rolling a die (1, 2, 3, 4, 5, or 6).

2. Continuous Distribution: Represents a random variable that can take any value within a range. It uses a probability density function (PDF).

```python
#Question 4: What is a binomial distribution, and how is it used in probability?
```

- A binomial distribution is a probability distribution that calculates the probability of getting a specific number of successes in a fixed number of independent trials, where each trial has only two possible outcomes: success or failure.

```python
#Question 5: What is the standard normal distribution, and why is it important?
```

- The standard normal distribution is a normal distribution with a mean of 0 and a standard deviation of 1. It is symmetrical and bell-shaped.
Importance:
1. It helps calculate probabilities and compare values from different datasets.
2. It is used to calculate z-scores and identify how far a value is from the mean.
3. It is useful in hypothesis testing and statistical analysis.

```python
#Question 6: What is the Central Limit Theorem (CLT), and why is it critical in statistics?
```

-The Central Limit Theorem (CLT) states that, for a sufficiently large sample size, the sampling distribution of the sample mean approaches a normal distribution, regardless of the population's original distribution, under suitable conditions.
